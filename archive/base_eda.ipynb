```python
import numpy as np
from PIL import Image
```

```python
def stretch(a, lo=2, hi=98):
    '''Percentile-stretch each channel of an HWC array into [0, 1]; NaNs become 0.'''
    out = np.zeros(a.shape, dtype=np.float32)
    print(out)
    for i in range(a.shape[-1]):
        print(i)
        ch = a[..., i]
        print(ch)
        vmin, vmax = np.nanpercentile(ch, [lo, hi])
        print(vmin , vmax)
        scaled = (ch - vmin) / (vmax - vmin + 1e-8)
        print(scaled)
        out[..., i] = np.nan_to_num(np.clip(scaled, 0, 1), nan=0.0)
        print(out)

    return out
```

```python
data  = np.load(r'C:\Users\moham\Downloads\2026_geoai_arctic_challenge_v1.0\competition_release\train\images\train_000001.npz')

print(f"keys in npz file: {data.files}")

image_array = data["image"]

rgb_raw = image_array[... , [0,1,2]]
rgb_stretched =stretch(rgb_raw)


rgb_8bit = (rgb_stretched * 255).astype(np.uint8)


img = Image.fromarray(rgb_8bit.astype("uint8"))
img.save("train_000001.jpg")
```
