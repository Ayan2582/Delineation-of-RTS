```python
# class for Patch Embedding - Part 1 of the VIT architecture
class PatchEmbedding(nn.Module):
    def __init(self):
        super().__init__()
        self.embed = nn.Conv2d(in_channels=num_channels , out_channels=embed_dim , kernel_size=patch_size , stride = patch_size)
        # Using conv2d cuz first of all it will convert the image into patches similar to how sentences are broken down into words and turned into vector embeddings , similarly we are converting each patch into its own patch_embedding
        # First stage outputt of conv2d is (Batch_size , channels , Height , Width)

    def forward(self , x):
        x = self.patch_embed(x)
        x = x.flatten(2)
        x = x.transpose(1,2)
        return x
    


```
